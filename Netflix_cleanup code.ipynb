```python
#import pandas
import pandas as pd

# Read in the dataset
Netflix_cleanup = pd.read_excel(r'C:\Users\Administrator\Documents\Custom Office Templates\Netflix cleanup.xltx')

#Drop/rename columns
Netflix_cleanup.columns

# 1. View summary information (column names, non-null counts, data types)
Netflix_cleanup.info()

# 2. View summary statistics for numerical
Netflix_cleanup.describe(include='all')

# 3. Display the first 5 rows of the dataset
Netflix_cleanup.head() 

#when a column value is numeric and you want to fill in the null values, you can fill it in with a -1
# Netflix_cleanup['Column_Name'] = Netflix_cleanup['Column_Name'].fillna(-1)
#To check which columns have null values, you can use the following code:
#null_columns = Netflix_cleanup[Netflix_cleanup['director'].isna()]
Netflix_cleanup[Netflix_cleanup['director'].isna()]
Netflix_cleanup[Netflix_cleanup['country'].isna()]


#4. Drop rows with null values in the 'director' and 'country' columns
Netflix_cleanup = Netflix_cleanup.dropna(subset=['director', 'country'])
#To replace the null values in the 'cast' column with a placeholder value, you can use the following code:
Netflix_cleanup['cast'] = Netflix_cleanup['cast'].fillna('Unknown')
# see the result of dropping the rows with null values
Netflix_cleanup.info()
Netflix_cleanup.head() 

#5. To ensure data types are correct for each column, you can use the following code:
Netflix_cleanup.dtypes
Netflix_cleanup.head() 
#fixing the data types for the 'release_year' and 'date_added' columns and duration column
Netflix_cleanup['release_year'] = Netflix_cleanup['release_year'].astype(int)
Netflix_cleanup['date_added'] = pd.to_datetime(Netflix_cleanup['date_added'], errors='coerce')
#Extract the number and convert it to numeric (int/float)
Netflix_cleanup['duration_num'] = (Netflix_cleanup['duration'].str.extract(r'(\d+)').astype('Int64') )
#Extract the unit (min, season, seasons) as text
Netflix_cleanup['duration_unit'] = (Netflix_cleanup['duration'].str.extract(r'([a-zA-Z]+)'))
# Drop the original object column (optional)
Netflix_cleanup.drop(columns=['duration'], inplace=True)

Netflix_cleanup.dtypes
Netflix_cleanup.head()

#6. Export to a CSV file
Netflix_cleanup.to_csv('Netflix_cleaned.csv', index=False) 

```
